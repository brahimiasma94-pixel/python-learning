```python
import numpy as np

X = np.array([5,7,20])
Y = np.array([9,15,4])
d=np.sqrt(np.sum((X-Y)**2))
print(f"Distance between X and Y is :{d}")
```

```python
import numpy as np
A = np.array([3,5,6,2,8,10,20,50])
first = A[0]
last = A[-1]
first_four=A[0:4]
two_five= A[2:5]
print(f"\nthe first four element is :{first_four},\nthe last element is :{last},\nthe element from index 2 to 5 excluding index 5 is :{two_five}")
A[:4]=7
print(f"the first element is : {first},\nthe matrix A before replacement is :{A}.")
```
